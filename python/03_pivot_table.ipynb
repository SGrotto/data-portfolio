```python
### Task 3: reviews by month × year (target: 12 min)

# Question: How many reviews were posted per calendar month, for each year from 2019 to 2026?

import pandas as pd 
display(pd.read_csv("../data/raw/reviews.csv", nrows=5)) # Check first five rows for column names 
df_review = pd.read_csv(
    "../data/raw/reviews.csv",
    usecols=["date","listing_id"],
    parse_dates=["date"])   # make date column in a date format directly   

df_review["year"]=df_review["date"].dt.year # Create year column just so the mask works without errors (otherwise filter in the .loc directly but not for now)
mask = ((df_review["year"]>=2019) & (df_review["year"]<=2026))

table=(df_review
.assign(year=lambda d: d["date"].dt.year) 
.assign(month=lambda d: d["date"].dt.month)
.loc[mask]
.pivot_table(
index="month",
columns="year",
values="listing_id",
aggfunc="count",)
)

# Sanity checks 
assert table.sum().sum() == mask.sum()
```
