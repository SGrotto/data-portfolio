```python
### Reviewing melt 

# df.melt(
#     id_vars=["month"],        # columns that stay as they are (the identifiers)
#     value_vars=[...],         # columns to stack (default: all the others)
#     var_name="year",          # name of the new column that holds the old column names
#     value_name="n_reviews",   # name of the new column that holds the values
# )
#   Optional: add .reset_index() if in the wide table the id_vars has no index 

# Take file 03's code 
import pandas as pd 
df_review = pd.read_csv(
    "../data/raw/reviews.csv",
    usecols=["date","listing_id"],
    parse_dates=["date"])   # make date column in a date format directly   

df_review["year"]=df_review["date"].dt.year # Create year column just so the mask works without errors (otherwise filter in the .loc directly but not for now)
mask = ((df_review["year"]>=2019) & (df_review["year"]<=2026))

table=(df_review
.assign(year=lambda d: d["date"].dt.year) 
.assign(month=lambda d: d["date"].dt.month)
.loc[mask]
.pivot_table(
index="month",
columns="year",
values="listing_id",
aggfunc="count",)
)
display(table)
display(table.dtypes)

# 1. Turn table into a long DataFrame long with three columns: month, year, n_reviews.
long=(table.reset_index()
    .melt(
    id_vars=["month"],
    var_name="year",
    value_name="n_reviews"
)
)
assert len(long)==96
assert long["n_reviews"].isna().sum()==3
long=long.dropna(subset=["n_reviews"]).astype({"n_reviews": int})
assert len(long)==93
display(long)

# 2. Sanity checks (as code):
# long has 12 × 8 = 96 rows. How many have a missing n_reviews, and which ones are they?
display(long["n_reviews"].isna().sum()) # 3 
display(long.loc[long["n_reviews"].isna()]) # last three months of 2026
# The sum of n_reviews equals table.sum().sum().
assert table.sum().sum() == long["n_reviews"].sum()

# 3. Decide: should the NaN rows stay in the long table? Drop them or keep them, with a one-line comment explaining why. 
# (Long tables don’t need placeholders the way a wide grid does.)
# Drop them: their row do not hold any useful information other than they are missing, so it's safe to remove them but add a note somewhere explaining that data for the last three months of 2026 is not yet available

# 4. Use the long format: compute the total number of reviews per year with groupby (expect 8 rows), and add a comment on which year is incomplete.
display(
    long 
    .groupby("year", as_index=False)["n_reviews"]
    .sum()
)
    # Important note: 2026 is incomplete: it is missing October-December, as well as the second-half of September, due to data not yet available

# 20 minutes
```
