```python
# Question: For each column of listings_clean.parquet, how many values are missing, and what share of rows is that?

import pandas as pd 
df = pd.read_parquet("../data/clean/listings_clean.parquet")

display(df.head(5))

# 1- Using df.
audit=(
    df 
    .isna().sum()
    .to_frame("n_missing") # Up to here we get the column n_missing of missings for each var
    .assign(pct_missing = lambda d: (100*d["n_missing"]/len(df)).round(1)) # .assign enables adding more columns (col1=..., col2=..., etc)
    .sort_values("n_missing", ascending=False)
)
display(audit)

# 2. Using DataFrame
audit2 = pd.DataFrame(
    {"n_missing": df.isna().sum(), "pct_missing": (df.isna().mean() * 100).round(1)}
).sort_values("n_missing", ascending=False)
display(audit2)

# Sanity checks 
# id has 0 missing values
assert df["id"].isna().sum() == 0

# for one column, n_missing + df[col].count() == len(df).
assert audit.loc["price_eur", "n_missing"]+df["price_eur"].notna().sum() == len(df)

# check the two ways yield the same result 
assert audit.equals(audit2)
```
